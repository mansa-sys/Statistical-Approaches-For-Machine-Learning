```python
#ASSIGNMENT 1 (Statistical Approaches)

#mean
class_intervals = [(15,25), (25,35), (35,45), (45,55), (55,65)] #age groups
frequency = [350, 375, 120, 80, 90] #number of persons

midpoints = [(low + high)/2 for (low, high) in class_intervals]
fx = [f * x for f, x in zip(frequency, midpoints)]

mean = sum(fx) / sum(frequency)
print("MEAN =", mean)
```

```python
# median
class_intervals = [(15,25), (25,35), (35,45), (45,55), (55,65)] #age groups
frequency = [350, 375, 120, 80, 90] #number of persons

# cumulative frequency
cf = []
total = 0
for f in frequency:
    total += f
    cf.append(total)

N = sum(frequency)
N2 = N / 2

# median class
for i in range(len(cf)):
    if cf[i] >= N2:
        median_class_index = i
        break


L = class_intervals[median_class_index][0]         # (lower boundary of median class)
h = class_intervals[median_class_index][1] - class_intervals[median_class_index][0]  # (class width)
f = frequency[median_class_index]                  # (frequency of median class)
CF = cf[median_class_index - 1] if median_class_index > 0 else 0  # (cumulative before median class)

median = L + ((N2 - CF) / f) * h
print("MEDIAN =", median)
```

```python
#mode
class_intervals = [(15,25), (25,35), (35,45), (45,55), (55,65)] #age groups
frequency = [350, 375, 120, 80, 90] #number of persons

modal_index = frequency.index(max(frequency))

L = class_intervals[modal_index][0]
h = class_intervals[modal_index][1] - class_intervals[modal_index][0]
f1 = frequency[modal_index]
f0 = frequency[modal_index - 1] if modal_index > 0 else 0
f2 = frequency[modal_index + 1] if modal_index < len(frequency) - 1 else 0

mode = L + ((f1 - f0) / (2*f1 - f0 - f2)) * h
print("MODE =", mode)
```
