```python
#2 fair and unbiased dice are rolled simultaneously to get some outcomes during the experiment.
#Write a python program to compute number of outcomes and probability of getting both faces same.

import itertools

# Generate all possible outcomes 
dice = [1, 2, 3, 4, 5, 6]
outcomes = list(itertools.product(dice, repeat=2))

# Total outcomes
total_outcomes = len(outcomes)

# Outcomes where both faces are same
same_face_outcomes = [outcome for outcome in outcomes if outcome[0] == outcome[1]]

# Probability
probability = len(same_face_outcomes) / total_outcomes

print("All outcomes:", outcomes)
print("Total number of outcomes:", total_outcomes)
print("Number of outcomes with both faces same:", len(same_face_outcomes))
print("Probability of getting both faces same:", probability)
```
